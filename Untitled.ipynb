#### Step 1: Feature Grouping (Columns ko Segregate Karna)

```python
# 2 Indentitfier aur  direget ko  alag karein

id_col = 'ID'
target_col ='Status'

# Step 1: Feature Grouping (Columns ko Segregate Karna
numric_feature =[
      col for col  in df.select_dtypes(include=[np.number]).columns
     if col  not in [id_col,target_col,'year']
]

categorical_feature=df.select_dtypes(exclude=[np.number]).columns.tolist()
                                  
print(f"Total Numeric Features ({len(numric_feature)}): {numric_feature}")
print(f"Total Categorical Features ({len(categorical_feature)}): {categorical_feature}")
```

### Step 2: Dynamic Cleaning & Imputation (Handling Missing Values)

```python
# Cleaning logic ko fit-transform pattern me structure karein
clean_df = df.copy()

# A. Numeric Imputation using Median
for col in numric_feature:
    median_val = clean_df[col].median()
    clean_df[col] = clean_df[col].fillna(median_val)


# B. Categorical Imputation using Mode
for col in categorical_feature:
    mode_val = clean_df[col].mode()[0]
    clean_df[col] = clean_df[col].fillna(mode_val)
print("Remaining Null Values:", clean_df.isnull().sum().sum())
```

### Find  out  Whoose  Column has highly skewed ()
And  Transformation  

```python
# 1. Skewness check karein (> 0.75 threshold)
skew_threshold = 0.75
skew_values = clean_df[numric_feature].skew()
high_skew_cols = skew_values[skew_values.abs() > skew_threshold].index.tolist()

print("Columns requiring Log Transform:", high_skew_cols)

# 2. Automated Log Transformation (np.log1p safe rehta hai zero values ke liye)
for col in high_skew_cols:
    clean_df[col] = np.log1p(clean_df[col])
```

### Visualization  of  corr  vs  target__cols

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Target ke sath correlation
corr = clean_df[numric_feature + [target_col]].corr()
top_features = corr[target_col].abs().sort_values(ascending=False).head(8).index

# 2. Clean Masked Seaborn Heatmap
plt.figure(figsize=(8, 6))

mask = np.triu(np.ones_like(clean_df[top_features].corr(), dtype=bool))
sns.heatmap(clean_df[top_features].corr(), mask=mask, annot=True, cmap='Blues', fmt='.2f')
plt.title("Top Predictive Features Correlation with Default")
plt.tight_layout()
plt.show()
```

###  Preprocessing 

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Numeric Pipeline: Median Impute -> Scale
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical Pipeline: Mode Impute -> One-Hot Encode
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

# Full Preprocessor
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numric_feature),
        ('cat', categorical_transformer, categorical_feature)
    ]
)

# Features (X) aur Target (y) split
X = df[numric_feature + categorical_feature]
y = df[target_col]

# Transform the raw dataset
X_processed = preprocessor.fit_transform(X)
print("Processed Dataset Shape ready for ML:", X_processed.shape)
```

### Top Predictive Features Isolate Karna (Correlation Matrix)
40+ features ke clutter se bachne ke liye Target (Status) ke saath highest absolute correlation wale top features filter karein.
### In this  model have leakage and next  I have  fixed it   

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score

# 1. Train-Test Split (80% training, 20% testing)
X_train, X_test, y_train, y_test = train_test_split(
    X_processed, y, test_size=0.2, random_state=42, stratify=y
)

# 2. Model Train Karna
model = RandomForestClassifier(n_estimators=100, max_depth=12, random_state=42, n_jobs=-1)
model.fit(X_train, y_train)

# 3. Predictions aur Evaluation
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]

print("ROC-AUC Score:", roc_auc_score(y_test, y_prob))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

### Feature Importance Check
Apne Random Forest model se poochhiye ki usne kis column ko dekh kar faisla liya:

```python


# Dekhein model ne kis feature ko sabse zyada importance di
importances = pd.Series(model.feature_importances_)
print("Top 3 most used features by model:")
print(importances.sort_values(ascending=False).head(3))
```

### How  to see exact feature  of  importance_df

```python
# Preprocessor se exact 69 feature names nikaalein
feature_names = preprocessor.get_feature_names_out()

# Importance ke sath ek clean DataFrame banayein
importances_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': model.feature_importances_
}).sort_values(by='Importance', ascending=False)

print("Actual Top 5 Features driving the model:")
print(importances_df.head(5))
```

### Classifier based  models to predict banks  Loan is  defaulter or  not   

```python
# 1. Leaky columns define karein
leaky_cols = ['rate_of_interest', 'Interest_rate_spread', 'Upfront_charges']

# 2. Numeric features list ko update karein
clean_numeric_features = [col for col in numric_feature if col not in leaky_cols]

# 3. Preprocessor ko naye features ke sath dobara banayein
clean_preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, clean_numeric_features),
        ('cat', categorical_transformer, categorical_feature)
    ]
)

# 4. Features (X) update karein
X_clean = df[clean_numeric_features + categorical_feature]
y = df['Status']

# 5. Transform & Train-Test Split
X_processed_clean = clean_preprocessor.fit_transform(X_clean)

X_train, X_test, y_train, y_test = train_test_split(
    X_processed_clean, y, test_size=0.2, random_state=42, stratify=y
)

# 6. Model Train & Evaluate
clean_model = RandomForestClassifier(n_estimators=100, max_depth=12, random_state=42, n_jobs=-1)
clean_model.fit(X_train, y_train)

y_prob_clean = clean_model.predict_proba(X_test)[:, 1]
y_pred_clean = clean_model.predict(X_test)

print("Realistic ROC-AUC Score:", roc_auc_score(y_test, y_prob_clean))
print("\nClassification Report:\n", classification_report(y_test, y_pred_clean))
```

```python
import joblib


# Models folder banayein
model_dir = Path('models')
model_dir.mkdir(exist_ok=True)

# 1. Preprocessor aur Model dono save karein
joblib.dump(clean_preprocessor, model_dir / 'preprocessor.joblib')
joblib.dump(clean_model, model_dir / 'random_forest_model.joblib')

print("Model & Preprocessor successfully saved inside 'models/' folder!")
```
